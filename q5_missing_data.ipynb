```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Import utilities from Q3
from q3_data_utils import load_data, detect_missing, fill_missing

# Load the data
df = load_data('data/clinical_trial_raw.csv')
print(f"Loaded {len(df)} patients")

# Prewritten visualization function for missing data
def visualize_missing_data(missing_counts):
    """
    Create a bar chart of missing values by column.
    
    Args:
        missing_counts: pandas Series with missing value counts per column
    """
    plt.figure(figsize=(10, 6))
    missing_counts.plot(kind='bar')
    plt.title('Missing Values by Column')
    plt.xticks(rotation=45)
    plt.ylabel('Number of Missing Values')
    plt.tight_layout()
    plt.show()
```

## Part 2: Compare Imputation Strategies (6 points)

For the 'cholesterol_total' column (which has missing values):

1. Fill with mean using `fill_missing()` utility
2. Fill with median using `fill_missing()` utility  
3. Forward fill using pandas `.fillna(method='ffill')`
4. Compare the three strategies - create a summary table showing:
   - Original mean/median
   - Mean/median after each strategy
   - How many values were filled

```python
# TODO: Compare imputation strategies
#Fill with mean
df_mean = fill_missing(df, 'cholesterol_total', strategy='mean')

#Fill with median
df_median = fill_missing(df, 'cholesterol_total', strategy='median')

#Fill with forward fill
df_ffill = fill_missing(df, 'cholesterol_total', strategy='ffill')

#Value counts
def count_filled_values(df):
    filled_counts = df['cholesterol_total'].value_counts()
    return filled_counts
print(f"\\nSite counts: {count_filled_values(df)}")

#Create a summary table
print("\n=== SUMMARY OF STRATEGIES ===")
print(f"Original rows: {len(df)}")
print(f"Original mean/median missing: {df['cholesterol_total'].isnull().sum()}")
print(f"Cholesterol Mean: filled {df_mean['cholesterol_total'].isnull().sum()} missing values with mean")
print(f"Cholesterol Median: filled {df_median['cholesterol_total'].isnull().sum()} missing values with median")
print(f"Cholesterol: forward filled {df_ffill['cholesterol_total'].isnull().sum()} missing dates")
print(f"Dropped {len(df)} rows with cholesterol missing")
print(f"Filled value counts: filled {df['cholesterol_total'].value_counts()}")
pivot_table = pd.DataFrame({
    'Strategy': ['Original', 'Mean', 'Median', 'Forward Fill'],
    'Missing Filled': [df['cholesterol_total'].isnull().sum(),
                       df_mean['cholesterol_total'].isnull().sum(),
                       df_median['cholesterol_total'].isnull().sum(),
                       df_ffill['cholesterol_total'].isnull().sum()],
    'Filled Values':[
        df['cholesterol_total'].mean(),
        df_mean['cholesterol_total'].mean(),
        df_median['cholesterol_total'].mean(),
        df_ffill['cholesterol_total'].mean()
    ]
})
print(pivot_table)
```

## Part 3: Dropping Missing Data (3 points)

1. Drop rows where ANY column has missing data - how many rows remain?
2. Drop rows where specific columns have missing data (e.g., only 'age' or 'bmi')
3. Which approach loses less data?

```python
# TODO: Drop missing rows with different strategies
df_complete = df.dropna(how='any')
print(f"\nAfter dropping rows missing in any column: {len(df_complete)} rows remain")

#Drop rows where specific columns are missing data
df_specific = df.dropna(subset=['bmi'], how='any')
print(f"After dropping rows missing bmi: {len(df_specific)} rows remain")
```

## Part 4: Create Clean Dataset (3 points)

Apply your chosen strategy to create a clean dataset:
1. Choose appropriate imputation for numeric columns
2. Drop rows with missing critical values (e.g., patient_id, age)
3. Save to `output/q5_cleaned_data.csv`
4. Save a missing data report to `output/q5_missing_report.txt`

```python
# TODO: Create and save clean dataset
df_clean = df.copy()
print("Saved cleaned data to output/q5_cleaned_data.csv")

#Imputation for numeric columns
numeric_cols = ['age', 'bmi', 'systolic_bp', 'diastolic_bp', 'cholesterol_total',
                'cholesterol_hdl', 'cholesterol_ldl', 'glucose_fasting',
                'follow_up_months', 'adverse_events', 'adherence_pct']

for f in numeric_cols:
    missing = df_clean[f].isnull().sum()
    if missing > 0:
        df_clean = fill_missing(df_clean, f, strategy='mean')
        print(f"Filled {missing} missing values in {f}")

#Drop rows with any missing data
df_clean = df_clean.dropna(subset=['patient_id', 'age'], how='any')
print(f"Cleaned dataset has {len(df_clean)} rows after dropping missing data")

#Save dataset to CSV
df_clean.to_csv('output/q5_cleaned_data.csv', index=False)

#save missing data report
with open('output/q5_missing_report.txt', 'w') as f:
    f.write("Missing Data Report\n")
    f.write("=" * 50 + "\n\n")
    f.write(f"Original rows: {len(df)}\n")
    f.write(f"Clean rows: {len(df_clean)}\n")
    f.write(f"Rows removed: {len(df) - len(df_clean)}\n")
```
